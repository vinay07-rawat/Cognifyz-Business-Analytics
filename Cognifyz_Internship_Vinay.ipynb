```python
import pandas as pd

# Load the dataset
# Replace 'your_file_name.csv' with the exact name of your uploaded CSV file
df = pd.read_csv('Dataset copy.csv')

# Display basic information about the dataset (columns, data types, and entries)
print("--- Dataset Information ---\n")
print(df.info())

# Display the first 5 rows of the dataset to understand its structure
print("\n--- First 5 Rows of the Dataset ---\n")
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Count the total number of Male and Female participants
gender_counts = df['gender'].value_counts()

# Display the numerical count
print("--- Gender Distribution Count ---\n")
print(gender_counts.to_string())
print("\n")

# Create a professional Bar Chart for visual representation
plt.figure(figsize=(7, 5))
gender_counts.plot(kind='bar', color=['#3498db', '#e74c3c'], edgecolor='black')

# Add titles and labels
plt.title('Gender Distribution of Participants')
plt.xlabel('Gender')
plt.ylabel('Number of Participants')
plt.xticks(rotation=0)
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# Identify numerical columns in the dataset
numerical_cols = df.select_dtypes(include=['int64', 'float64'])

# Calculate Mean, Median, and Standard Deviation
# Rounding to 2 decimal places for a cleaner look
statistics = numerical_cols.agg(['mean', 'median', 'std']).round(2)

# Display the descriptive statistics (Transposed for better readability)
print("--- Descriptive Statistics for Numerical Columns ---\n")
print(statistics.T.to_string())
```

```python
import matplotlib.pyplot as plt

# Define the columns representing different investment avenues
investment_cols = ['Mutual_Funds', 'Equity_Market', 'Debentures',
                   'Government_Bonds', 'Fixed_Deposits', 'PPF', 'Gold']

# Calculate the mean (average) rank for each avenue and sort them in ascending order
# Lower numerical value means it is highly preferred by the participants
investment_means = df[investment_cols].mean().sort_values()

# Display the textual output
print("--- Average Rank of Investment Avenues ---")
print("(Note: Lower score indicates higher preference)\n")
print(investment_means.to_string())

# Identify and print the most preferred avenue (the one with the lowest mean)
most_preferred = investment_means.index[0]
print(f"\n=> The most preferred investment avenue is: {most_preferred}\n")

# Create a professional bar chart for visualization
plt.figure(figsize=(10, 6))
investment_means.plot(kind='bar', color='#f39c12', edgecolor='black')

# Add titles and labels
plt.title('Preference of Investment Avenues (Lower Score = Higher Preference)')
plt.xlabel('Investment Avenue')
plt.ylabel('Average Rank')
plt.xticks(rotation=45)
plt.tight_layout()

# Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Columns containing reasons for different investments
reason_columns = ['Reason_Equity', 'Reason_Mutual', 'Reason_Bonds', 'Reason_FD']

print("--- Summary of Reasons for Investment Choices ---\n")

# Loop through each reason column to print the summary
for col in reason_columns:
    # Extract investment name for a cleaner output
    investment_type = col.replace('Reason_', '')
    print(f"Top reasons for investing in {investment_type}:")
    print(df[col].value_counts().to_string())
    print("-" * 40)

# Visualizing reasons for investing in Mutual Funds as an example for the video
plt.figure(figsize=(9, 5))
sns.countplot(x='Reason_Mutual', data=df, palette='Set2', edgecolor='black')

# Add titles and labels
plt.title('Primary Reasons for Investing in Mutual Funds')
plt.xlabel('Reason')
plt.ylabel('Number of Participants')
plt.tight_layout()

# Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# The exact column name from the dataset that contains savings objectives
savings_col = 'What are your savings objectives?'

# Calculate the frequency of each savings objective
objective_counts = df[savings_col].value_counts()

# Display the textual summary
print("--- Main Savings Objectives Mentioned by Participants ---\n")
print(objective_counts.to_string())
print("\n")

# Create a professional Pie Chart for visual representation
plt.figure(figsize=(8, 8))
plt.pie(objective_counts, labels=objective_counts.index, autopct='%1.1f%%',
        startangle=140, colors=['#ff9999', '#66b3ff', '#99ff99'],
        wedgeprops={'edgecolor': 'black'})

# Add a title
plt.title('Primary Savings Objectives of Participants')
plt.tight_layout()

# Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Identify the column containing the sources of investment information
source_col = 'Source'

# Calculate the frequency of each information source
source_counts = df[source_col].value_counts()

# Display the textual summary
print("--- Common Sources of Investment Information ---\n")
print(source_counts.to_string())
print("\n")

# Create a professional horizontal bar chart for visualization
plt.figure(figsize=(10, 6))
sns.barplot(x=source_counts.values, y=source_counts.index, palette='viridis', edgecolor='black')

# Add titles and labels for the presentation
plt.title('Common Sources Participants Rely On For Investment Info')
plt.xlabel('Number of Participants')
plt.ylabel('Information Source')
plt.tight_layout()

# Display the chart
plt.show()
```
